Next, let's try to enhance the map by adding the launch outcomes for each site, and see which sites have high success rates.
Recall that data frame spacex_df has detailed launch records, and the `class` column indicates if this launch was successful or not


```python
# TASK 2: Mark success/failed launches for each site on the map

# Create a base map centered around the US (good starting point)
center_lat = spacex_df['Lat'].mean()
center_lon = spacex_df['Long'].mean()

launch_map = folium.Map(location=[center_lat, center_lon], zoom_start=5)

# Create a MarkerCluster
marker_cluster = MarkerCluster()

# Add markers for each launch
for index, record in spacex_df.iterrows():
    lat = record['Lat']
    lon = record['Long']
    site = record['Launch Site']
    outcome = record['class']          # 1 = Success, 0 = Failed
    
    if outcome == 1:
        color = 'green'
        icon = 'rocket'
        popup_text = f"Success at {site}<br>Payload: {record['Payload Mass (kg)']} kg"
    else:
        color = 'red'
        icon = 'remove'
        popup_text = f"Failed at {site}<br>Payload: {record['Payload Mass (kg)']} kg"
    
    folium.Marker(
        location=[lat, lon],
        popup=folium.Popup(popup_text, max_width=300),
        icon=folium.Icon(color=color, icon=icon, prefix='fa')
    ).add_to(marker_cluster)

# Add the cluster to the map
marker_cluster.add_to(launch_map)

# Display the map
launch_map
```

```python
# TASK 2: Mark success/failed launches with colored markers

# Create map centered on launch sites
site_map = folium.Map(location=[spacex_df['Lat'].mean(), spacex_df['Long'].mean()], zoom_start=5)

# Create MarkerCluster
marker_cluster = MarkerCluster()

# TODO: Create a new column for marker color
spacex_df['marker_color'] = spacex_df['class'].apply(lambda x: 'green' if x == 1 else 'red')

# Add markers to the cluster
for index, record in spacex_df.iterrows():
    marker = folium.Marker(
        location=[record['Lat'], record['Long']],
        popup=folium.Popup(f"""
            <b>Site:</b> {record['Launch Site']}<br>
            <b>Outcome:</b> {'Success' if record['class']==1 else 'Failed'}<br>
            <b>Payload:</b> {record['Payload Mass (kg)']} kg
        """, max_width=300),
        icon=folium.Icon(color='white', icon_color=record['marker_color'], icon='rocket', prefix='fa')
    )
    marker_cluster.add_child(marker)

# Add cluster to map
site_map.add_child(marker_cluster)

# Display map
site_map
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
# ==================== TASK 3: Distance to Proximities ====================

# Example coordinates for CCAFS LC-40 (most common launch site)
launch_site_lat = 28.562302
launch_site_lon = -80.577356

# 1. Closest Coastline
coastline_lat = 28.56367
coastline_lon = -80.57163

# 2. Closest Railway
railway_lat = 28.57215
railway_lon = -80.58543

# 3. Closest Highway
highway_lat = 28.55423
highway_lon = -80.59012

# 4. Closest City (Titusville)
city_lat = 28.6121
city_lon = -80.8075

# Distance function
def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0  # Earth radius in km
    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance = R * c
    return distance

# Calculate distances
dist_coast = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
dist_rail = calculate_distance(launch_site_lat, launch_site_lon, railway_lat, railway_lon)
dist_highway = calculate_distance(launch_site_lat, launch_site_lon, highway_lat, highway_lon)
dist_city = calculate_distance(launch_site_lat, launch_site_lon, city_lat, city_lon)

print(f"Distance to Coastline: {dist_coast:.2f} km")
print(f"Distance to Railway: {dist_rail:.2f} km")
print(f"Distance to Highway: {dist_highway:.2f} km")
print(f"Distance to City: {dist_city:.2f} km")
```
